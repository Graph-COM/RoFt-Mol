```python
import pandas as pd

dataset = "muv"
column_names = ['dataset', 'split', 'fewshot', 'seed', 'reg', 'score', 'std']
df = pd.read_csv('sup_largemix_results/{}/{}_log.csv'.format(dataset, dataset), names=column_names, header=None)
```

```python
seed = " All seed"
split = " Split:size"
fewshot_num = 50
reg_type = "bss"
reg = " Reg_"+ reg_type + "_"
fewshot = " Fewshot_False_{}".format(fewshot_num)
```

```python
# filtered_df = df[(df['seed'] == " All seed") & (df['split'] == split) & (df['fewshot'] == fewshot) & df['reg'].str.contains("Tune")]
# filtered_df = df[(df['seed'] == " All seed") & (df['fewshot'] == fewshot) & df['reg'].str.contains("LP-FT")]
# filtered_df = df[(df['seed'] == " All seed") & (df['split'] == split) & (df['fewshot'] == fewshot) & df['reg'].str.startswith(" WISE")]
# filtered_df = df[(df['seed'] == " All seed") & (df['split'] == split) & (df['fewshot'] == fewshot) & df['reg'].str.contains("FTMAP")]
filtered_df = df[(df['seed'] == " All seed") & (df['split'] == split) & df['reg'].str.contains(reg) & (df['fewshot'] == fewshot)]
# filtered_df = df[(df['seed'] == " All seed") & (df['fewshot'] == fewshot) & df['reg'].str.contains("DFTMAP")]
# filtered_df = filtered_df.iloc[[0, 3, 6, 9, 12]]
filtered_df
```

```python
scores = filtered_df['score'].values
stds = filtered_df['std'].values
scores, stds
```

```python
import numpy as np
import re
def extract_score(s):
    match = re.search(r'Best Acc:\s*(\d+\.\d+|\d+)', s)
    return float(match.group(1)) if match else None

def extract_std(s):
    match = re.search(r'std:\s*(\d+\.\d+|\d+)', s)
    return float(match.group(1)) if match else None

# Apply the function to each element in the numpy array
extracted_scores = np.array([extract_score(item) for item in scores])
extracted_stds = np.array([extract_std(item) for item in stds])
```
